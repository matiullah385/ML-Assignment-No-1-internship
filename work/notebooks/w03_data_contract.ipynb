## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

This dataset has several limitations:

It does not capture user intent or behavior beyond clicks and impressions.
Early data may be incomplete (e.g., only from Search Console), which can create imbalance.
Time windows may overlap or have gaps, affecting trend analysis.
External factors (like SEO changes, competition, or seasonality) are not included.
The dataset cannot fully explain why CTR changes, only what changed.
